# Neural Network V3

In this version I tried dropout (0, 0.2 and 0.5) and weight decay (0, 0.0001 and 0.001), one at a time. Dropout randomly turns off some neurons during training so the network can't depend too much on any one of them, and weight decay keeps the weights small. I also added early stopping, which stops training when the validation loss hasn't improved for a few epochs.

```python
from pathlib import Path
import subprocess, sys
import pandas as pd
from IPython.display import display

ROOT = Path.cwd()
if not (ROOT / 'run_experiments.py').is_file():
    ROOT = ROOT.parent
assert (ROOT / 'run_experiments.py').is_file(), 'Open this notebook from the project or notebooks folder.'
TASK = 'both'  # Change to 'image' or 'sentiment' to run only one task.
EPOCHS = 20
DEVICE = 'auto'  # or 'cpu' / 'cuda'
print(f'Project: {ROOT}; task: {TASK}; device: {DEVICE}')
```

## Training V1 to V3

I run V1 and V2 first so that V3 starts from the settings V2 picked on validation.

```python
subprocess.run([
    sys.executable, str(ROOT / 'run_experiments.py'),
    '--task', TASK, '--version', '3', '--epochs', str(EPOCHS), '--device', DEVICE,
], cwd=ROOT, check=True)
```

## V3 results

```python
summary = pd.read_csv(ROOT / 'results' / 'nn' / 'nn_final_summary.csv')
display(summary.loc[summary.version == 3])
from IPython.display import Image
for task in (['image', 'sentiment'] if TASK == 'both' else [TASK]):
    display(pd.read_csv(ROOT / 'results' / task / 'nn_v3_validation.csv').head())
    display(Image(filename=str(ROOT / 'results' / 'nn' / f'nn_v3_{task}_sweeps.png')))
submission = ROOT / 'submissions' / 'submission_v3.csv'
if submission.is_file():
    print(f'Submission saved to {submission}')
    display(pd.read_csv(submission).head())
```
