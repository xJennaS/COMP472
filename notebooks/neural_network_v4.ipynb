# Neural Network V4

Here I changed the size of the network. First I tried two hidden layers with 128, 256, 512 and 1024 neurons each. Then I tried 1, 2 and 3 hidden layers of 256 neurons. I changed one of these at a time.

```python
from pathlib import Path
import subprocess, sys
import pandas as pd
from IPython.display import display

ROOT = Path.cwd()
if not (ROOT / 'run_experiments.py').is_file():
    ROOT = ROOT.parent
assert (ROOT / 'run_experiments.py').is_file(), 'Open this notebook from the project or notebooks folder.'
TASK = 'both'  # Change to 'image' or 'sentiment' to run only one task.
EPOCHS = 20
DEVICE = 'auto'  # or 'cpu' / 'cuda'
print(f'Project: {ROOT}; task: {TASK}; device: {DEVICE}')
```

## Training V1 to V4

I run V1 to V3 first so V4 starts from the settings picked in V3. This version also saves plots of the validation accuracy for each setting.

```python
subprocess.run([
    sys.executable, str(ROOT / 'run_experiments.py'),
    '--task', TASK, '--version', '4', '--epochs', str(EPOCHS), '--device', DEVICE,
], cwd=ROOT, check=True)
```

## V4 results

```python
summary = pd.read_csv(ROOT / 'results' / 'nn' / 'nn_final_summary.csv')
display(summary.loc[summary.version == 4])
from IPython.display import Image
for task in (['image', 'sentiment'] if TASK == 'both' else [TASK]):
    display(pd.read_csv(ROOT / 'results' / task / 'nn_v4_validation.csv').head())
    display(Image(filename=str(ROOT / 'results' / 'nn' / f'nn_v4_{task}_sweeps.png')))
submission = ROOT / 'submission' / 'nn' / 'submission_v4_nn.csv'
if submission.is_file():
    print(f'Submission saved to {submission}')
    display(pd.read_csv(submission).head())
```
