# Neural Network V1

This is my baseline network. It has two hidden layers (256 and 64 neurons) with ReLU, and I trained it with SGD with momentum and a fixed learning rate. I didn't use any regularization yet. I used the same train and validation split as the rest of the project (seed 42). If both tasks are run, it also makes `submission/nn/submission_v1_nn.csv`.

```python
from pathlib import Path
import subprocess, sys
import pandas as pd
from IPython.display import display

ROOT = Path.cwd()
if not (ROOT / 'run_experiments.py').is_file():
    ROOT = ROOT.parent
assert (ROOT / 'run_experiments.py').is_file(), 'Open this notebook from the project or notebooks folder.'
TASK = 'both'  # Change to 'image' or 'sentiment' to run only one task.
EPOCHS = 20
DEVICE = 'auto'  # or 'cpu' / 'cuda'
print(f'Project: {ROOT}; task: {TASK}; device: {DEVICE}')
```

## Training V1

This only runs V1. It saves a summary file and a validation file for each task, in the same format as the k-NN results.

```python
subprocess.run([
    sys.executable, str(ROOT / 'run_experiments.py'),
    '--task', TASK, '--version', '1', '--epochs', str(EPOCHS), '--device', DEVICE,
], cwd=ROOT, check=True)
```

## V1 results

```python
summary = pd.read_csv(ROOT / 'results' / 'nn' / 'nn_final_summary.csv')
display(summary.loc[summary.version == 1])
for task in (['image', 'sentiment'] if TASK == 'both' else [TASK]):
    print(f'{task}: v1 validation predictions')
    display(pd.read_csv(ROOT / 'results' / task / 'nn_v1_validation.csv').head())
submission = ROOT / 'submissions' / 'submission_v1.csv'
if submission.is_file():
    print(f'Submission saved to {submission}')
    display(pd.read_csv(submission).head())
```
